<b>2.2) Pretratamiento de Missing</b>

```r
# Creamos funciones auxiliares que detecten el nivel de missing por variables

def var_percent(df):
    cols_null_vals = df.isnull().sum(axis = 0)
    cols_null_vals = (100 * cols_null_vals) / df.shape[0]
    cols_null_vals = cols_null_vals[cols_null_vals != 0]
    cols_nulls = cols_null_vals.sort_values(ascending = False)
    if len(cols_nulls) > 0:
        print('Porcentaje de Nulos por Variable:')
        print(cols_nulls)
    else:
        print('No hay nulos por variable')

# Creamos esta función auxiliar que detecta el nivel total de missing en toda la base de datos

def null_percent(df):
    total_nulls = (100 * df.isnull().sum(). sum()) / (df.shape[0] * df.shape[1])
    print(f'Porcentaje Total de Nulos = {total_nulls}')
```

<b>2.3.2) Outliers influyentes para el modelo</b>

Para el tratamiento de outliers, en este caso vamos a aplicar un método simple y robusto para tratar el problema de outliers. Vamos a aplicar una ***winsorización al 5%-95%***. Para ellos vamos a tratar a las variables continuas con la siguiente transformación:

$
X_i^{win} =
\begin{cases}
q_{0.05}, & \text{si } X_i < q_{0.05} \\
X_i, & \text{si } q_{0.05} \le X_i \le q_{0.95} \\
q_{0.95}, & \text{si } X_i > q_{0.95}
\end{cases}
$

Con esto podremos **mitigar de manera robusta el efecto de valores extremos** sobre modelos sensibles a outliers (como los modelos de regresión tradicionales).

Para nuestro caso vamos a aplicar esta transformación a las variables continuas. Las variables categóricas se mantienen en sus valores originales. Es importante remarcar que esta winsorización, por ser una transformación, se debe aplicar en la muestra de entrenamiento y luego aplicar esta misma a la muestra de test. ¿Por qué?

```r
from sklearn.model_selection import train_test_split

X = df.drop(columns=["SalePrice"])
y = df["SalePrice"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=123)
```

```r
# Definimos las variables a las cuales aplicaremos la winsorización (aquellas que son float64) en la muestra de train

winsor_vars = ["LotFrontage", "LotArea", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2",
    "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF", "LowQualFinSF", "GrLivArea",
    "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch", "3SsnPorch", "ScreenPorch", "PoolArea", "MiscVal"]

quantiles = {}

for v in winsor_vars:
    q05 = X_train[v].quantile(0.05)
    q95 = X_train[v].quantile(0.95)
    quantiles[v] = (q05, q95)

    X_train[v] = X_train[v].clip(q05, q95)
```

```r
# Aplicamos esta misma transformación en la muestra de test

for v in winsor_vars:
    q05, q95 = quantiles[v]
    X_test[v] = X_test[v].clip(q05, q95)
```

### <b>3) Proceso de Feature Engineering</b>

Para el proceso inicial y


```r
def feature_engineering(df):
    df = df.copy()

    # A) Transformaciones log para variables muy asimétricas
    df["log_GrLivArea"] = np.log(df["GrLivArea"])
    df["log_TotalBsmtSF"] = np.log(df["TotalBsmtSF"] + 1)  # evitar log(0)

    # B) Cantidad de baños totales de la casa
    df["total_bath"] = (
        df["FullBath"] +
        0.5 * df["HalfBath"] +
        df["BsmtFullBath"] +
        0.5 * df["BsmtHalfBath"]
    )

    # C) Edad de la casa y remodelación
    df["house_age"] = df["YrSold"] - df["YearBuilt"]
    df["remod_age"] = df["YrSold"] - df["YearRemod/Add"]

    # D) Espacio total finalizado
    df["total_finished_sf"] = df["GrLivArea"] + df["TotalBsmtSF"]
    df["log_total_finished_sf"] = np.log(df["total_finished_sf"])

    return df
```

## 5.1. Modelo $\mathcal{M}_1$: Regresión Lasso

```r
install.packages('glmnet')
library(glmnet)
```

```r
# Creamos una función para calcular métricas de performance de regresión:

calcular_errores <- function(y_test, y_pred) {
  # Asegurarse de que son vectores numéricos
  y_test <- as.numeric(y_test)
  y_pred <- as.numeric(y_pred)

  # MAE: Mean Absolute Error
  mae <- mean(abs(y_test - y_pred))

  # MAPE: Mean Absolute Percentage Error
  mape <- mean(abs((y_test - y_pred) / ifelse(y_test == 0, NA, y_test)), na.rm = TRUE) * 100 # Evitamos división por cero

  # RMSE: Root Mean Squared Error
  rmse <- sqrt(mean((y_test - y_pred)^2))

  # R^2
  r2 <- 1 - sum((y_test - y_pred)^2) / sum((y_test - mean(y_test))^2)

  # Crear resumen
  resumen <- data.frame(
    MAE = mae,
    MAPE = mape,
    RMSE = rmse,
    R2 = r2
  )

  return(resumen)
}
```

```r
# Definimos la fórmula y preparamos los objetos tipo matriz para usar glmnet

x_train <- model.matrix(~ ., data = train_data_scaled)[, -1]
x_test  <- model.matrix(~ ., data = test_data_scaled)[, -1]

y_train <- as.matrix(y_train)
y_test <- as.matrix(y_test)
```

```r
head(x_train)
```

```r
# 1. Entrenamiento con validación cruzada k-fold (k = 5) para Lasso

set.seed(123)

cv_lasso <- cv.glmnet(
  x = x_train,
  y = y_train,
  alpha = 1,          # LASSO (L1)
  nfolds = 5,         # k = 5 folds
  standardize = FALSE  # La data ya está estandarizada
)
```

```r
# Mejor lambda
lambda_opt <- cv_lasso$lambda.min
lambda_opt
```

```r
# 2. Entrenamos el modelo final usando lambda óptimo

lasso_opt <- glmnet(
  x = x_train,
  y = y_train,
  alpha = 1,
  lambda = lambda_opt,
  standardize = FALSE
)
```

## 5.3. Modelo $\mathcal{M}_3$: Regresión Elastic Net

```r
## 1. Entrenamos un modelo Elastic Net:

# Valores de alpha que vamos a probar
alpha_values <- seq(0.1, 0.9, by = 0.1)

# Guardar resultados
cv_results <- list()
cv_errors <- c()

set.seed(123)  # reproducibilidad

for(a in alpha_values) {
  cv_model <- cv.glmnet(
    x = x_train,
    y = y_train,
    alpha = a,
    nfolds = 5,
    standardize = FALSE
  )
  cv_results[[as.character(a)]] <- cv_model
  cv_errors <- c(cv_errors, min(cv_model$cvm))  # error mínimo CV
}
```

```r
# Mejor alpha
best_alpha <- alpha_values[which.min(cv_errors)]
best_alpha

# Mejor lambda correspondiente al alpha óptimo
best_lambda <- cv_results[[as.character(best_alpha)]]$lambda.min
best_lambda
```

```r
# 2. Entrenarmos el modelo final Elastic Net con los mejores hiperparámetros

elastic_model <- glmnet(
  x = x_train,
  y = y_train,
  alpha = best_alpha,
  lambda = best_lambda,
  standardize = FALSE
)
```

```r
# 3. Evaluación final de test (25%)

M3_pred <- predict(elastic_model, newx = x_test)

M3_error = calcular_errores(y_test, M3_pred)
M3_error
```
